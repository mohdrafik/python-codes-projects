```python
class Complex(object):
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary
        
    def __add__(self, no):
        return Complex(self.real + no.real, self.imaginary + no.imaginary)

    def __sub__(self, no):
        return Complex(self.real - no.real, self.imaginary - no.imaginary)

    def __mul__(self, no):
        realpart = self.real * no.real - self.imaginary * no.imaginary
        imaginarypart = self.real * no.imaginary + no.real * self.imaginary
        return Complex(realpart, imaginarypart)

    def __truediv__(self, no):
        modulusComplexNumber = (no.real)**2 + (no.imaginary)**2
        realpart = (self.real * no.real + self.imaginary * no.imaginary) / modulusComplexNumber
        imaginarypart = (self.imaginary * no.real - self.real * no.imaginary) / modulusComplexNumber
        return Complex(realpart, imaginarypart)

    def mod(self):
        """Returns the modulus as a complex number with imaginary part 0"""
        return Complex(((self.real)**2 + (self.imaginary)**2) ** 0.5, 0.00)

    def __str__(self):
        """Formats the complex number output properly"""
        if self.imaginary == 0:
            result = "%.2f+0.00i" % (self.real)
        elif self.real == 0:
            if self.imaginary >= 0:
                result = "0.00+%.2fi" % (self.imaginary)
            else:
                result = "0.00-%.2fi" % (abs(self.imaginary))
        elif self.imaginary > 0:
            result = "%.2f+%.2fi" % (self.real, self.imaginary)
        else:
            result = "%.2f-%.2fi" % (self.real, abs(self.imaginary))
        return result
# 
# ✅ Fixed indentation in `if __name__ == '__main__':`
# if __name__ == '__main__':
    # c = map(float, input().split())
    # d = map(float, input().split())
    # x = Complex(*c)  # Unpacking real & imaginary
    # y = Complex(*d)
# 
    # ✅ Ensuring __str__() is used
    # print(x+y)  
    # print(x-y)  
    # print(x*y)  
    # print(x/y)  
    # print(x.mod())  
    # print(y.mod())  
```

# example for practice :
-  "%.2fi" % (self.imaginary)	Inserts self.imaginary, rounded to 2 decimal places, into the string.

-  "%.2f+%.2fi" % (self.real, self.imaginary)	Inserts both self.real and self.imaginary into the formatted string.

-  "0.00-%.2fi" % (abs(self.imaginary))	Ensures the imaginary part is always positive, manually handling the - sign.
